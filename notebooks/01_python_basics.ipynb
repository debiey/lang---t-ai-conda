## 1. Variables and Basic Data Types

## 2. Lists (very important for AI data)

```python
scores = [85, 90, 78, 92, 88]
print(scores)
print("First score:", scores[0])
print("Last score:", scores[-1])
print("Number of scores:", len(scores))
```

## 3. Dictionaries

```python
student = {
    "name": "Chioma",
    "age": 17,
    "city": "Lagos",
    "interested_in_ai": True
}

print(student)
print("Name:", student["name"])
print("City:", student["city"])
```

## 4. Conditions (if / else)

```python
score = 72

if score >= 70:
    print("Excellent!")
elif score >= 50:
    print("Good effort")
else:
    print("Keep practicing")
```

## 5. Loops

```python
fruits = ["mango", "orange", "banana", "pineapple"]

for fruit in fruits:
    print("I like", fruit)
```

```python
# Loop with numbers
for i in range(5):
    print("Number:", i)
```

## 6. Functions

```python
def greet(name):
    return "Hello " + name + "! Welcome to the AI Lab."

message = greet("Tunde")
print(message)
```

```python
def average(numbers):
    return sum(numbers) / len(numbers)

scores = [80, 90, 70, 85]
print("Average score:", average(scores))
```
